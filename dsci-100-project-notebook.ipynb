#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).

```r
# count missing values in each column we use, in the raw data
# is.na() gives TRUE for a missing value, sum() counts the TRUEs
missing_table <- march_madness |>
  summarize(
    seed_missing = sum(is.na(seed)),
    reg_wins_pct_missing = sum(is.na(reg_wins_pct)),
    tourney_wins_missing = sum(is.na(tourney_wins)),
    bid_missing = sum(is.na(bid))
  )

missing_table
```

```r
# Plot 1: distribution of tournament wins
plot_wins <- march_tidy |>
  ggplot(aes(x = tourney_wins)) +
  geom_histogram(binwidth = 1) +
  labs(x = "Number of Tournament Wins",
       y = "Number of Teams",
       title = "Distribution of Tournament Wins")

plot_wins
```

```r
# Plot 2: proportion of each type of bid
bid_proportions <- march_tidy |>
  group_by(bid) |>
  summarize(count = n()) |>
  mutate(proportion = count / sum(count))

plot_bid <- bid_proportions |>
  ggplot(aes(x = bid, y = proportion)) +
  geom_bar(stat = "identity") +
  labs(x = "Type of Bid",
       y = "Proportion of Teams",
       title = "Proportion of Teams by Type of Bid")

plot_bid
```

Visualization1: This plot shows the average number of tournament wins for each seed. Seed is one of our predictors, so this helps us see if it is useful for predicting tournament wins. Teams with a lower seed number (stronger teams) win more games on average, so seed should be a helpful predictor in our model.

```r
seed_wins <- march_tidy |>
  group_by(seed) |>
  summarize(mean_tourney_wins = mean(tourney_wins))

plot_seed <- seed_wins |>
  ggplot(aes(x = seed, y = mean_tourney_wins)) +
  geom_point() +
  geom_line() +
  labs(x = "Seed (1 = strongest)",
       y = "Average Number of Tournament Wins",
       title = "Average Tournament Wins by Seed")

plot_seed
```

Visualization 2: This plot shows how regional win percentage relates to the number of tournament wins. Regional win percentage is our second predictor. Teams with a higher win percentage tend to win more tournament games, so it should also help our prediction. The colours show the type of bid. Most teams with many tournament wins have a high win percentage. Many teams have 0 or 1 wins, which may make it harder to predict teams with many wins.

```r
plot_pct <- march_tidy |>
  ggplot(aes(x = reg_wins_pct, y = tourney_wins, colour = bid)) +
  geom_point(alpha = 0.4) +
  labs(x = "Regional Win Percentage",
       y = "Number of Tournament Wins",
       colour = "Type of Bid",
       title = "Regional Win Percentage vs Tournament Wins")

plot_pct
```

I plan to use K-nearest neighbors (KNN) regression to predict the number of tournament wins from a team’s seed and regional win percentage. This method is appropriate because my response variable, tourney_wins, is numeric, and KNN regression does not assume a straight-line relationship between the predictors and the response. It predicts a team’s wins by taking the average wins of the K most similar teams in the training data. One limitation is that KNN does not give an equation, so it is hard to explain how much each predictor matters. It also predicts poorly outside the range of the training data. In addition, most teams win only 0 or 1 games, so the model may under-predict wins for strong teams. 

Before building the model, I will split the data once at the start: 75% for training and 25% for testing, using set.seed() so my results can be reproduced. The test set will only be used at the end. Because KNN uses distance, I will standardize (center and scale) both predictors in a recipe so that one variable does not have more influence just because of its larger values. Instead of a separate validation set, I will use 5-fold cross-validation on the training set to choose the K with the lowest RMSPE. Then I will fit the model with this K on the whole training set and calculate RMSPE on the test set. RMSPE is appropriate because it shows the average size of my prediction error in the same units as the response (number of wins).
